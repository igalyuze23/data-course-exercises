```python
def move_zeros(L):
    L1 = []
    
    for i in L:
        if i != 0:
            L1.append(i)
    
    zero_n = len(L) - len(L1)
    
    for i in range(zero_n):
        L1.append(0)
        
    return L1

my_list = [0, 5, 0, 12, 0, 7]
print(move_zeros(my_list))
```

```python
def sum_divisors(num):
    total = 0
    for i in range(1, num):
        if num%i==0: 
            total = total + i
            
    return total
print("Finding perfect numbers...")
perfect_numbers = []
for i in range(1,10000):
    if i ==sum_divisors(i):
        perfect_numbers.append(i)
print(perfect_numbers)
```

```python
def factorial (n):
    if n==1:
        return 1
    else:
        return n*factorial(n-1)
print(f"the factorial is: {factorial(5)}")
```
